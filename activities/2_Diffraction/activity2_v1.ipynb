# Activity 2: Diffraction-limit

A telescope's aperture diffracts light: even a perfect point source is smeared into an
**Airy pattern** of angular size $\theta \approx 1.22\,\lambda/D$. Two sources closer
than this blur together into one — no matter how good the optics are.

**In this activity you will:**
1. Compute $\theta$ for different wavelengths and dish sizes.
2. Use sliders to *see* two point sources merge as they get closer than $\theta$.

## Diffraction limit

$\theta = 1.22\,\lambda/D$ (Rayleigh criterion).

## When do two sources become unresolved?

* **$\lambda$ dropdown** — pick visible light, 21 cm radio, etc.
* **$D$ slider** (log scale, mm → 1000 km) — aperture diameter.
* **sep / $\theta$ slider** — separation of the two sources in units of the
  diffraction limit.

When the sep/$\theta \approx 1$ the two peaks merge — that's the **Rayleigh criterion**.
