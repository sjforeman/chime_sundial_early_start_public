# Activity 2: Diffraction limit

A telescope's aperture (i.e. the limited area over which it collects light) causes *diffraction*: even a perfect point source is smeared into an
**Airy pattern** of angular size $\theta \approx 1.22\,\lambda/D$. This is known as the *Rayleigh criterion*.

If two sources are closer than this, they blur together into one when the telescope observes the sky.

**In this activity you will:**
1. Compute $\theta$ for different wavelengths and dish sizes.
2. Use sliders to *see* two point sources merge as they get closer than $\theta$.




## Diagram controls

* **$\lambda$ dropdown** - allows you to pick from several categories of light, each with a different wavelength $\lambda$
* **$D$ slider** - aperture diameter from mm to 1000 km (on a log scale - small changes to the slider have larger effects as $D$ gets larger).
  * NOTE: The above two sliders don't change the diagram, but do change the angles involved - pay attention to how $\theta$ changes.
* **"Sep. / $\theta$" slider** - separation of the two sources divided by the Rayleigh criterion size ($1.22\,\lambda/D$).

When the "Sep. / $\theta$" $\approx 1$, the Airy patterns merge.
