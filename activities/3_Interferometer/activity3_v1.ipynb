3. Slide $\theta$ slowly and watch the red dot ride the fringe — that oscillation is exactly what a correlator records as a source drifts overhead.

## From Two Antenna Signals to Visibility
Visibility measures the correlation between the signals received by two antennas:
**multiply antenna 1's voltage by antenna 2's voltage and average over time**.

Below, antenna 2's signal is a phase-shifted copy of antenna 1's signal, with $\phi = \frac{2\pi B\sin\theta}{\lambda}$.

Slide $\phi$ and watch:

- the two sine waves shift relative to each other,
- their instantaneous product oscillates (grey),
- its **time average** (red line) is $\tfrac{1}{2}\cos\phi$.

In phase → maximum positive correlation.  
Half a wavelength out → maximum negative correlation.  
Quarter of a wavelength out → zero correlation.
