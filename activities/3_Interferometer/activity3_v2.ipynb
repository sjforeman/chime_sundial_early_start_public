# Activity 3: Two-element interferometer

Move the sliders and watch two things at once:

* **Top panel** — the geometry: two antennas, the incoming wavefront, and the extra path length $\Delta L$ the wavefront needs to travel to reach one antenna after the other.
* **Bottom panel** — the interference pattern of the waves measured by the two antennas, as a function of source position (represented as the angle $\theta$ away from the point directly above the telescope).

## Activity
* Observe how the Airy patterns move as you vary the "Sep. / $\theta$" slides.
* For different values of $\lambda$, what value of $D$ do you need to obtain an angular resolution $\theta$ of $\sim$ 1 arcsec? (This is roughly what you can achieve with a backyard optical telescope.)

**Try this:**
1. As you increase the antenna distance $D$, notice that the interference pattern oscillates faster and faster as a function of $\theta$.
2. As you increase the light wavelength $\lambda$, the interference pattern oscillates more and more slowly as a function of $\theta$. 

(Points 1 and 2 reflect the fact that the interference pattern depends on the ratio $\lambda/D$, just like the angular resolution from the diffraction activity.)

3. Vary $\theta$ slowly and watch the red dot on the interference pattern. As an astronomical object (such as a galaxy) moves across the sky, the antenna pair measures a different point on the interference pattern. This provides some information about where the source is.
