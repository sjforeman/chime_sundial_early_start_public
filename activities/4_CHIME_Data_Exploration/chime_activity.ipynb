# Warm-up

First, let's go through the format of the plots in this notebook.

- ***Blue points***: Oscillatory part of a *visibility* measured by CHIME, which is the product of the signals measured by each antenna in a *baseline* (a pair of antennas with a certain displacement between them). The plot's title shows the east-west separation of the antennas. When the plot is zoomed all the way out, the blue points blend together into a single line, if you zoom in enough, you will see that there are separate points, connected by a blue line to guide the eye.
- ***Black points***: Absolute value of the same visibility.

- ***Upper x axis***: time of day when the visibility was measured. The time axis spans 2 days, which are indicated in the plot's title.
- ***Lower x axis***: right ascension (RA) corresponding to each time. Right ascension is measured in degrees and is an azimuthal angle (like a longitude) used to specify a location on the sky. Right ascension doesn't change as the Earth turns; instead, it's used to refer to the locations of specific celestial objects (e.g. a certain galaxy). The value on the lower x axis is the right ascension corresponding to the point that is directly over the telescope at the time on the upper x axis.
- ***Y axis***: value of the visibility, in Jy (Janskys, a unit of brightness common in radio astronomy).

The example plotting command below can be executed by selecting it and typing Ctrl-Enter. The options control select a specific day and baseline, and also allow you to adjust the limits of the x (RA) and y axes. Finally, the `figsize` argument can be adjusted to make the plot larger or smaller.

## The Sun

The Sun is extremely bright ($\sim$ 200kJy) at frequencies measured by CHIME. In scientific analysis of CHIME data, any data taken when the Sun is above the horizon is considered to be contaminated, but in this notebook we have included data at these times. Because of this, you will need to make sure you don't mistake the Sun or its associated contamination for one of the objects in the table above.

Here are some facts about the Sun:

1. On each day, the Sun passes over CHIME at a similar local time, but at a different value of right ascension. (This is because of the difference in length between a "solar day", which is our standard definition of a day, and a "sidereal day", which is how long it takes for distant astrophysical objects to come back into view at the same location on the sky.)

2. The Sun is above the horizon (and has a strong effect on CHIME data) for between 8 and 16 hours each day, staying up longer in the summer than in the winter.
